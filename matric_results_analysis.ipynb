```python
import pandas as pd
import numpy as np
```

```python
def clean_district_name(name):
    """Fixes reversed and multi-line district names (e.g., 'TSAE\nOZN\nDERFLA' -> 'ALFRED NZO EAST')"""
    if pd.isna(name):
        return name
    parts = str(name).split('\n')
    # Reverse the character order of each word and then reverse the order of the words
    reversed_parts = [p[::-1].strip() for p in parts]
    return " ".join(reversed_parts[::-1]).strip()
```

```python
def process_esc_data(file_path):
    # Load the data - column names are likely just indices '0', '1', etc. due to headers
    df = pd.read_csv(/content/ESC_Matric_Data_2024.csv,/content/FS_Matric_Data_2024.csv)

    # Slice from row 3 where the actual data starts
    data = df.iloc[3:].copy()

    # Clean district names and forward-fill them
    data['0'] = data['0'].apply(clean_district_name).ffill()

    records = []
    for _, row in data.iterrows():
        # Skip rows without an EMIS number (header artifacts or empty rows)
        if pd.isna(row['1']): continue

        # Basic school info
        base_info = {
            'Province': 'Eastern Cape',
            'District': row['0'],
            'EMIS No': row['1'],
            'Centre No': row['3'],
            'Centre Name': row['5'],
            'Quintile': row['12']
        }

        # Map yearly columns (Column indices based on ESC file structure)
        # 2022
        records.append({**base_info, 'Year': 2022, 'No Progressed': row['14'],
                        'Total Wrote': row['17'], 'Total Achieved': row['18'], 'Achieved %': row['19']})
        # 2023
        records.append({**base_info, 'Year': 2023, 'No Progressed': row['20'],
                        'Total Wrote': row['21'], 'Total Achieved': row['23'], 'Achieved %': row['24']})
        # 2024
        records.append({**base_info, 'Year': 2024, 'No Progressed': row['25'],
                        'Total Wrote': row['26'], 'Total Achieved': row['28'], 'Achieved %': row['29']})

    return pd.DataFrame(records)

def process_fs_data(file_path):
    df = pd.read_csv(file_path)
    data = df.iloc[3:].copy()
    data['0'] = data['0'].apply(clean_district_name).ffill()

    records = []
    for _, row in data.iterrows():
        if pd.isna(row['1']): continue

        base_info = {
            'Province': 'Free State',
            'District': row['0'],
            'EMIS No': row['1'],
            'Centre No': row['2'],
            'Centre Name': row['3'],
            'Quintile': row['9']
        }

        # Map yearly columns (Column indices based on FS file structure)
        # 2022
        records.append({**base_info, 'Year': 2022, 'No Progressed': row['11'],
                        'Total Wrote': row['12'], 'Total Achieved': row['14'], 'Achieved %': row['17']})
        # 2023
        records.append({**base_info, 'Year': 2023, 'No Progressed': row['19'],
                        'Total Wrote': row['20'], 'Total Achieved': row['23'], 'Achieved %': row['24']})
        # 2024
        records.append({**base_info, 'Year': 2024, 'No Progressed': row['26'],
                        'Total Wrote': row['29'], 'Total Achieved': row['30'], 'Achieved %': row['31']})

    return pd.DataFrame(records)
```

```python
esc_clean = process_esc_data('ESC_Matric_Data_2024.csv')
fs_clean = process_fs_data('FS_Matric_Data_2024.csv')

# Combine them into one master dataframe
master_df = pd.concat([esc_clean, fs_clean], ignore_index=True)

# Save the final cleaned data
master_df.to_csv('Cleaned_Matric_Data_2024.csv', index=False)

print("Processing complete. Saved to 'Cleaned_Matric_Data_2024.csv'")
```
